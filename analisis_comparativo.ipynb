# ¿Cuál modelo tuvo menor latencia promedio?

De acuerdo a lo que se evidencia en el tracking de cada uno de los modelos, podemos decir que el tiene mayor latencia es de modelo de Ollama. Esto podría deberse a la cantidad de recursos que requiere el modelo para procesar una solicitud, debido a que en teoria, al ser local debería ahorrar tiempo en comparación a hacer una solicitud a un modelo en la nube.

# ¿Cuál generó más tokens?

Apesar de que Gemini tuvo una latencia mucho más baja que Ollama, el numero de Tokens generados fue mayor, es decir que su respuesta fue mucho más detallada. Esto podría estar relacionado al prompt pero aún así es un resultado prometedor.

# ¿Qué parámetros podrías optimizar en futuros experimentos?

Para futuros experimentos, podríamos ajustar la temperatura para entregar una respuesta mucho más precisa aunque podría significar mucho más costo. Se podría mejorar el prompt para que se más claro y directo, para que así el modelo use menos tokens y responda más rápido. También se podrían probar versiones más ligeras o rápidas (“flash”, “mini”, etc.) puede reducir latencia del modelo.
